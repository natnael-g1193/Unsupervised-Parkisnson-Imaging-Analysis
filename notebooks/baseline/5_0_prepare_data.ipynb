```python
# Import Libraries
import pandas as pd
```

```python
# Files paths
# DATA_VALIDATION_PATH = "../../data/baseline/final_validation_combined_vae_data_beta_9.00e-05.csv"
# DATA_TRAIN_PATH = "../../data/baseline/final_train_combined_vae_data_beta_9.00e-05.csv"

DATA_VALIDATION_PATH = "../../data/baseline/final_validation_combined_vae_data.csv"
DATA_TRAIN_PATH = "../../data/baseline/final_train_combined_vae_data.csv"
```

```python
# Load the CSV file into a DataFrame
df = pd.read_csv(DATA_VALIDATION_PATH)

print("Shape:", df.shape)
print("Unique patients:", df['PATNO'].nunique())
print("Total rows:", len(df))
print("Rows per patient (mean):", len(df) / df['PATNO'].nunique())
print("\nLabel counts:\n", df['label'].value_counts())
print("\nColumns:\n", df.columns.tolist())
print("filtered columns:", df.columns[~df.columns.str.startswith("latent")])
```

```python
# Load the training data
df_train = pd.read_csv(DATA_TRAIN_PATH)

print("Shape:", df_train.shape)
print("Unique patients:", df_train['PATNO'].nunique())
print("Total rows:", len(df_train))
print("Rows per patient (mean):", len(df_train) / df_train['PATNO'].nunique())
print("\nLabel counts:\n", df_train['label'].value_counts())
print("\nEVENT_ID counts:\n", df_train['EVENT_ID'].value_counts().head(10))

# Check the patient-level split
val_patients  = set(pd.read_csv(DATA_VALIDATION_PATH)['PATNO'])
train_patients = set(df_train['PATNO'])
overlap = train_patients & val_patients
print(f"\nPatient overlap between train and val: {len(overlap)}")
```
